## 1.2 Основы языка Python


### Переменные и числовые типы


````python
# Целое, вещественное и комплексное числа имеют разные типы.
# 整数、浮点数和复数具有不同的数据类型。
a = 5
b = 2.0
z = 3 + 4j

print(type(a))
print(type(b))
print(type(z))
````

````python
# Сравниваем обычное деление, целую часть частного, остаток и степень.
# 比较普通除法、向下取整除法、余数和幂运算。
a, b = 5, 2.0
print(a+b, a-b, a*b, a/b, a//b, a%b, a**b)
````

### Строки


````python
# Оператор + соединяет строки; пробел уже включён в s1.
# + 用于拼接字符串；s1 已包含末尾空格。
s1 = "hello "
s2 = 'world'
s = s1 + s2
````

### Булевы значения


````python
# Логические значения True и False используются в условиях.
# 布尔值 True 和 False 用于条件判断。
x = True
y = False
````

````python
# Задаём числовые значения для примеров сравнений и условий.
# 设置用于比较和条件判断示例的数值。
a = 2
b = 3
````

````python
# Выполняется только та ветвь, условие которой истинно.
# 根据条件是否成立，只执行对应的分支。
age = 20
if age >= 18:
    print("Совершеннолетний")
else:
    print("Несовершеннолетний")
````

### Список и кортеж


````python
# Индексы начинаются с нуля; правая граница среза не включается.
# 索引从 0 开始；切片不包含右端点。
a = [1, 2, 3, 4]
print(a[0])     # 1
print(a[1:3])   # [2, 3]
````

````python
# Список изменяем: заменяем первый элемент и добавляем новый в конец.
# 列表可修改：替换首元素，并在末尾添加新元素。
a[0] = 10
a.append(5)
print(a)  # [10, 2, 3, 4, 5]
````

````python
# Кортеж поддерживает индексы и срезы, как и список.
# 元组与列表一样支持索引和切片。
b = (1, 2, 3, 4)
print(b[0])     # 1
print(b[1:3])   # (2, 3)
````

````python
# Элементы кортежа неизменяемы; перехватываем ожидаемую ошибку.
# 元组元素不能重新赋值；捕获预期的类型错误。
try:
    b[0] = 10   # TypeError
except TypeError as error:
    print("Ожидаемая ошибка TypeError:", error)
````

### Словарь


````python
# Словарь связывает уникальные ключи с их значениями.
# 字典将唯一的键与对应的值关联。
dic = {'a': 1, 'b': 2, 'c': 3}
````

````python
# Добавляем ключ d, меняем значение a и удаляем ключ b.
# 添加键 d，修改 a 的值，并删除键 b。
dic['d'] = 4
dic['a'] = 10
del dic['b']
````

### Пользовательская функция: закон Фурье


````python
# Закон Фурье: k — теплопроводность, dT/L — градиент температуры.
# 傅里叶定律：k 为热导率，dT/L 为温度梯度。
def heat_flux(k, dT, L):
    # Знак минус: поток направлен против градиента температуры.
    # 负号表示热流方向与温度梯度相反。
    q = -k * dT / L
    return q

q = heat_flux(150, 20, 0.01)
print(q)
````

### Анонимные функции


````python
# lambda задаёт короткую функцию без отдельного блока def.
# lambda 用于定义简短函数，无需单独编写 def 代码块。
square = lambda x: x ** 2
print(square(5))
````

````python
# map применяет функцию к каждому элементу; list собирает результаты.
# map 对每个元素应用函数；list 将结果收集成列表。
nums = [1, 2, 3, 4]
squared = list(map(
    lambda x: x ** 2, nums))
print(squared)
````

### Цикл for: перебор значений


````python
# Цикл for последовательно перебирает значения списка.
# for 循环依次遍历列表中的值。
temperatures = [280, 300, 320]
for T in temperatures:
    print(T)
````

````python
# items() возвращает пары ключ–значение словаря.
# items() 返回字典中的键值对。
T = {'A': 280, 'B': 300}
for point, value in T.items():
    print(point, value)
````

### Циклы: while, break, continue


````python
# Увеличиваем счётчик, чтобы условие while со временем стало ложным.
# 递增计数器，使 while 条件最终不成立。
i = 0
while i < 5:
    print(i)
    i += 1
````

````python
# break немедленно завершает весь цикл при i == 2.
# 当 i == 2 时，break 立即结束整个循环。
for i in range(5):
    if i == 2:
        break
    print(i)
````

````python
# continue пропускает только текущую итерацию при i == 2.
# 当 i == 2 时，continue 仅跳过本次迭代。
for i in range(5):
    if i == 2:
        continue
    print(i)
````

## 1.3 Научные вычисления с NumPy


### Сложение списков и массивов NumPy


````python
# Создаём два списка для сравнения с массивами NumPy.
# 创建两个列表，用于与 NumPy 数组比较。
import numpy as np
a = [1, 2, 3, 4]
b = [2, 3, 4, 5]
````

````python
# Для списков + означает соединение, а не поэлементное сложение.
# 列表的 + 表示拼接，不表示逐元素相加。
print(a + b)
````

````python
# Для массивов + складывает элементы; скаляр прибавляется ко всем элементам.
# 数组的 + 逐元素相加；标量会加到每个元素上。
x = np.array(a)
y = np.array(b)
print(x + y)
print(x + 1)
````

### Тип элементов массива


````python
# Тип int не хранит дробную часть: при заполнении 2.5 превращается в 2.
# int 类型不能保存小数部分，填充时 2.5 会转为 2。
a = np.zeros(4, dtype=int)
a.fill(2.5)
print(a)
````

````python
# Преобразуем массив в float, чтобы сохранить дробные значения.
# 将数组转为 float，以保存小数值。
b = a.astype(float)
b.fill(2.5)
print(b)
````

### Форма массива и его оси


````python
# shape — размеры осей, ndim — число осей, size — число элементов.
# shape 是各轴长度，ndim 是维数，size 是元素总数。
a = np.array([[1, 2, 3],
              [4, 5, 6]])
print(a.shape)
print(a.ndim)
print(a.size)
````

### Трансляция массивов: прибавляем строку к матрице


````python
# Трансляция прибавляет вектор b к каждой строке матрицы a.
# 广播机制将向量 b 加到矩阵 a 的每一行。
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([1, 2, 3])
print(a + b)
````

### Равномерная сетка: шаг и число точек


````python
# arange задаёт шаг 2; верхняя граница 10 не включается.
# arange 指定步长 2；不包含上界 10。
x = np.arange(0, 10, 2)
print(x)
````

````python
# linspace задаёт число точек; обе границы включены по умолчанию.
# linspace 指定点数；默认包含两端点。
x = np.linspace(0, 10, 6)
print(x)
````

### Массивы из списковых выражений


````python
# Списковое включение вычисляет 1 + i для каждого i.
# 列表推导式为每个 i 计算 1 + i。
a = np.array([
    1 + i for i in range(10)
])
print(a)
````

````python
# Возводим каждое значение 1 + i в квадрат перед созданием массива.
# 先将每个 1 + i 平方，再构造数组。
a = np.array([
    (1 + i)**2 for i in range(10)
])
````

````python
# Внешний цикл задаёт строки, внутренний — элементы каждой строки.
# 外层循环生成各行，内层循环生成行内元素。
a = np.array([
    [i * j for j in range(4)]
    for i in range(3)
])
````

### Случайные числа: два распределения


````python
# Фиксированное зерно позволяет повторить последовательность случайных чисел.
# 固定随机种子可重现随机数序列。
rng = np.random.default_rng(1)
````

````python
# Получаем 10 равномерных случайных чисел из интервала [0, 1).
# 从 [0, 1) 均匀分布抽取 10 个随机数。
a = rng.random(10)
````

````python
# Получаем 10 значений нормального распределения со средним 0 и σ = 1.
# 从均值为 0、标准差为 1 的正态分布抽取 10 个值。
b = rng.normal(0, 1, size=10)
````

### Случайные целые числа и выбор элементов


````python
# Повторная инициализация с тем же зерном начинает последовательность заново.
# 用相同种子重新初始化，会从头生成同一随机序列。
rng = np.random.default_rng(1)
````

````python
# Генерируем шесть целых чисел от 1 до 19 включительно.
# 生成 6 个整数，取值范围为 1 到 19。
a = rng.integers(1, 20, size=6)
print(a)
````

````python
# replace=False запрещает выбирать один элемент списка повторно.
# replace=False 表示不放回抽样，同一元素不能重复抽取。
values = [10, 20, 30, 40]
b = rng.choice(values, size=3,
               replace=False)
print(b)
````

### Индексация и срезы: одномерный массив


````python
# Создаём одномерный массив для примеров индексации.
# 创建一维数组，用于演示索引。
a = np.array([1, 2, 3, 4, 5, 6])
````

````python
# Отрицательный индекс отсчитывается с конца; ::2 берёт каждый второй элемент.
# 负索引从末尾计数；::2 每隔一个元素取值。
print(a[1:3])   # [2 3]
print(a[1:-2])  # [2 3 4]
print(a[::2])   # [1 3 5]
````

````python
# Присваивание по индексу изменяет исходный массив.
# 通过索引赋值会修改原数组。
a[0] = 10
print(a)
# [10  2  3  4  5  6]
````

### Индексация: строки и столбцы


````python
# Первый индекс выбирает строку, второй — столбец; : выбирает всю ось.
# 第一个索引选行，第二个选列；: 选择该轴的全部元素。
a = np.array([[0, 1, 2, 3],
              [10, 11, 12, 13]])
print(a[1, 3])
print(a[1])
print(a[:, 1])
````

````python
# Изменяем элемент во второй строке и четвёртом столбце.
# 修改第二行、第四列的元素。
a[1, 3] = -1
# [[ 0  1  2  3]
#  [10 11 12 -1]]
````

### Логическая маска: выбор по условию


````python
# Булева маска оставляет только элементы, для которых условие истинно.
# 布尔掩码只保留满足条件的元素。
a = np.arange(0, 10)
mask = (a > 5)
b = a[mask]
print(b)  # [6 7 8 9]
````

### Срез массива и независимая копия


````python
# Срез обычно разделяет память с исходным массивом; меняется и a.
# 切片通常与原数组共享内存，因此修改 b 也会影响 a。
a = np.array([0, 1, 2, 3, 4])
b = a[2:4]
b[0] = 10
print(a)
````

````python
# copy() создаёт независимые данные; изменение b не затрагивает a.
# copy() 创建独立数据，修改 b 不会影响 a。
a = np.array([0, 1, 2, 3, 4])
b = a[2:4].copy()
b[0] = 10
print(a)
````

### Объединение массивов и выбор оси


````python
# Оба массива имеют форму (2, 3): ось 0 — строки, ось 1 — столбцы.
# 两个数组的形状均为 (2, 3)：轴 0 对应行，轴 1 对应列。
x = np.array([[0, 1, 2], [10, 11, 12]])
y = np.array([[50, 51, 52], [60, 61, 62]])
````

### Изменение формы и транспонирование


````python
# reshape меняет форму, сохраняя число элементов и порядок данных.
# reshape 改变形状，保持元素总数和数据顺序。
a = np.arange(6)
b = a.reshape((2, 3))
print(a)
print(b)
````

````python
# Транспонирование меняет местами строки и столбцы.
# 转置将行和列互换。
print(b.T)
````

### Сортировка: новый массив или изменение на месте


````python
# Неотсортированный массив содержит и повторяющиеся значения.
# 待排序数组中包含重复值。
data = np.array([3, 1, 4, 1, 5])
````

````python
# np.sort возвращает новый массив; исходный data остаётся прежним.
# np.sort 返回新数组，原数组 data 保持不变。
sorted_data = np.sort(data)
print(sorted_data)
print(data)
````

````python
# Метод sort() сортирует исходный массив на месте.
# sort() 方法直接对原数组进行原地排序。
data.sort()
print(data)
````

### Скалярное произведение векторов


````python
# Скалярное произведение — сумма попарных произведений элементов.
# 点积等于对应元素乘积之和。
A = np.array([1, 2, 3])
B = np.array([1, 100, 10000])
print(np.dot(A, B))  # 30201
````

### Матричное произведение


````python
# Для A @ B число столбцов A должно совпадать с числом строк B.
# 计算 A @ B 时，A 的列数必须等于 B 的行数。
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[1, 2], [3, 4], [5, 6]])
print(A @ B)
````

## 1.4 Визуализация с Matplotlib


### Первый график в Matplotlib


````python
# Вычисляем sin(x) на сетке и строим график с подписями осей.
# 在网格上计算 sin(x)，并绘制带坐标轴标签的曲线。
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
````

### Погрешности: errorbar


````python
# dT задаёт погрешности температуры; yerr отображает их вертикальными отрезками.
# dT 给出温度误差，yerr 将其显示为竖直误差棒。
t = np.array([0, 10, 20, 30, 40])
T = np.array([80, 65, 54, 46, 39])
dT = np.array([2, 3, 2, 2, 3])
plt.figure(figsize=(6, 4))
plt.errorbar(t, T, yerr=dT,
             fmt='o', capsize=5)
plt.xlabel('t, с', fontsize=16)
plt.ylabel('T, °C', fontsize=16)
plt.tick_params(labelsize=14)
plt.grid()
plt.tight_layout()
plt.show()
````

### Несколько кривых и легенда


````python
# label задаёт подписи кривых, которые затем выводит legend().
# label 设置曲线名称，legend() 显示图例。
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')

plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()
````

### Оформление и сохранение графика


````python
# Сохраняем рисунок в curves.png с разрешением 300 dpi до вызова show().
# 在调用 show() 前，以 300 dpi 将图形保存为 curves.png。
x = np.linspace(0, 2 * np.pi, 100)
plt.figure(figsize=(6, 4))
plt.plot(x, np.sin(x), 'r-', lw=2,
         label='sin(x)')
plt.plot(x, np.cos(x), 'k--', lw=2,
         label='cos(x)')
plt.xlabel('x', fontsize=14)
plt.ylabel('y', fontsize=14)
plt.legend()
plt.savefig('curves.png', dpi=300,
            bbox_inches='tight')
plt.show()
````

### Точки и линии с маркерами


````python
# scatter показывает отдельные точки, plot соединяет точки линией.
# scatter 绘制散点，plot 用线连接各点。
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure(figsize=(6, 4), dpi=100)
plt.scatter(x, y1, color='blue',
            label='sin(x)')
plt.plot(x, y2, color='green',
         marker='o', linestyle='-',
         linewidth=2, label='cos(x)')
plt.xlabel('x', fontsize=18)
plt.ylabel('y', fontsize=18)
plt.tick_params(labelsize=16)
plt.legend(fontsize=14)
plt.grid()
plt.show()
````

### Гистограмма


````python
# Гистограмма группирует 1000 наблюдений в 30 интервалов.
# 直方图将 1000 个样本划分到 30 个区间中。
rng = np.random.default_rng(1)
data = rng.normal(0, 1, size=1000)
plt.rc('font', family='DejaVu Sans')

plt.figure(figsize=(6, 4))
plt.hist(data, bins=30,
         color='skyblue',
         edgecolor='black')
plt.xlabel('Значение')
plt.ylabel('Число наблюдений')
plt.tight_layout()
plt.show()
````

### Координатная сетка: meshgrid


````python
# meshgrid создаёт таблицы координат для всех сочетаний x и y.
# meshgrid 为 x 和 y 的所有组合生成坐标网格。
x = np.array([0, 1, 2])
y = np.array([0, 1])
X, Y = np.meshgrid(x, y)
````

### Одно поле на плоскости и в пространстве


````python
# Вычисляем скалярное поле Z в каждой точке двумерной сетки.
# 在二维网格的每个点上计算标量场 Z。
x = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, x)
Z = np.sin(np.sqrt(X**2 + Y**2))
````

````python
# contourf показывает значения поля цветом; colorbar связывает цвет и число.
# contourf 用颜色表示场值，colorbar 给出颜色与数值的对应关系。
plt.figure()
c = plt.contourf(X, Y, Z, levels=50)
plt.colorbar(c, label='z')
plt.show()
````

````python
# Те же данные Z изображаем как высоту трёхмерной поверхности.
# 将同一组 Z 数据作为高度绘制三维曲面。
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.show()
````

### Несколько графиков: subplots


````python
# subplots создаёт две области рисования с общей осью x.
# subplots 创建两个共享 x 轴的子图。
x = np.linspace(0, 2 * np.pi, 100)
fig, axes = plt.subplots(2, 1,
    figsize=(6, 5), sharex=True)
axes[0].plot(x, np.sin(x))
axes[0].set_ylabel('sin(x)',
                   fontsize=14)
axes[1].plot(x, np.cos(x), color='red')
axes[1].set_ylabel('cos(x)',
                   fontsize=14)
axes[1].set_xlabel('x', fontsize=14)
for ax in axes:
    ax.tick_params(labelsize=12)
    ax.grid()
fig.tight_layout()
plt.show()
````

### Логарифмическая шкала: semilogy


````python
# На полулогарифмическом графике экспоненциальное затухание выглядит прямой.
# 在半对数坐标中，指数衰减表现为直线。
t = np.linspace(0, 100, 200)
dT = 200 * np.exp(-0.05 * t)
fig, axes = plt.subplots(2, 1,
    figsize=(6, 5), sharex=True)
axes[0].plot(t, dT)
axes[1].semilogy(t, dT)
for ax in axes:
    ax.set_ylabel('T - T_env, K',
                  fontsize=12)
    ax.tick_params(labelsize=12)
    ax.grid()
axes[1].set_xlabel('t, с', fontsize=14)
fig.tight_layout()
plt.show()
````

### Умножение в NumPy и MATLAB

Поэлементное умножение: `A * B` в NumPy, `A .* B` в MATLAB.
Матричное произведение: `A @ B` в NumPy, `A * B` в MATLAB.


````python
# Сравниваем поэлементное, матричное, скалярное и внешнее произведения.
# 比较逐元素乘法、矩阵乘法、点积和外积。
import numpy as np
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(A * B)
print(A @ B)
print(np.dot(a, b))
print(np.outer(a, b))
````
