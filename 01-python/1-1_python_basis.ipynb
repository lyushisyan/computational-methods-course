## 2. Структуры данных (数据结构)

### 2.1 Список: list (列表)

Список — изменяемая последовательность. Можно хранить разные типы данных и повторяющиеся значения. Команды ниже выполняются по порядку.

列表是可变的有序序列，可以存放不同类型的数据和重复值。下面的命令按顺序执行。

```python
lst = [1, 2, 3, 4]
```

```python
lst.append(5)  # добавить в конец (添加到末尾)
lst
```

```python
lst.insert(1, 10)  # вставить на позицию (插入至某处)
lst
```

```python
lst.remove(3)  # Удалить первое значение 3 (删除首次出现的值 3)
lst
```

```python
lst.pop(3)  # Удалить элемент с индексом 3 и вернуть его (删除索引为 3 的元素并返回它)
```

```python
lst # Список после удаления элементов (删除元素后的列表)
```

```python
lst[1:3]   # срез от i до j (не включая j) (从 i 到 j（不包括 j）的切片)
```

`lst + 1` вызывает `TypeError`: оператор `+` соединяет списки, а не прибавляет число к каждому элементу.

`lst + 1` 会引发 `TypeError`：`+` 用于连接列表，不会把一个数加到每个元素上。

### 2.2 Кортеж: tuple (元组)

Кортеж — неизменяемая последовательность. Доступ по индексу и срезы работают так же, как у списка.

元组是不可变的有序序列，支持与列表相同的索引和切片操作。

```python
tpl = (1, 2, 3, 4)
```

```python
tpl[0]  # доступ к элементу по индексу (通过索引访问元素)
```

```python
tpl[1:3]  # срез (подкортеж) (切片（子元组）)
```

```python
len(tpl)  # длина кортежа (元组长度)
```

```python
3 in tpl  # проверка наличия элемента (检查元素是否存在)
```

`tpl.append(5)` вызывает `AttributeError`: у кортежа нет метода `append()`. Нельзя заменить, добавить или удалить элемент самого кортежа.

`tpl.append(5)` 会引发 `AttributeError`：元组没有 `append()` 方法。不能替换、添加或删除元组自身的元素。

### 2.3 Множество: set (集合)

Множество хранит уникальные элементы. Порядок не гарантирован, доступа по индексу нет. Элементы должны быть хешируемыми: например, числа или строки. Пустое множество создаётся как `set()`, а `{}` создаёт словарь.

集合保存不重复的元素，不保证顺序，不支持按索引访问。元素必须可哈希，例如数值或字符串。空集合用 `set()` 创建，`{}` 创建的是空字典。

```python
s = {1, 2, 3, 4}
```

```python
s.add(5)  # добавить элемент (增加元素)
s
```

```python
s.remove(3)  # удалить элемент (删除元素)
s
```

```python
s1 = {1, 2, 3}
s2 = {3, 4, 5}
```

```python
s1 | s2  # объединение (并集)
```

```python
s1 & s2  # пересечение (交集)
```

```python
s1 - s2  # разность (差集)
```

### 2.4 Словарь: dict (字典)

Словарь хранит пары «ключ–значение». Ключи уникальны, значения могут повторяться. Сохраняется порядок добавления ключей.

字典保存“键—值”对。键唯一，值可以重复；保留键的插入顺序。

```python
dic = {'a': 1, 'b': 2, 'c': 3}
```

```python
dic['b']  # получить значение по ключу (通过键获取值)
```

```python
dic['d'] = 4  # добавить значение (添加值)
dic
```

```python
dic['a'] = 10  # изменить значение (修改值)
dic
```

```python
del dic['b']  # удалить пару (删除键值对)
dic
```

```python
dic.keys()  # все ключи (所有键)
```

```python
dic.values()  # все значения (所有值)
```

```python
dic.items()  # все пары ключ-значение (所有键值对)
```

## 3. Функции (函数)

### 3.1 Пользовательская функция: закон Фурье (自定义函数：傅立叶定律)

Для плоского слоя с линейным профилем температуры:

$$q=-k\frac{\Delta T}{L},\qquad \Delta T=T_2-T_1.$$

`k` — теплопроводность, Вт/(м·К); `dT` — разность температур, К; `L` — толщина, м; `q` — плотность теплового потока, Вт/м². Знак минус задаёт перенос тепла в сторону меньшей температуры.

`def` определяет функцию; `return` возвращает результат.

对于温度线性分布的平板层：`k` 是导热系数，单位 W/(m·K)；`dT` 是温差，单位 K；`L` 是厚度，单位 m；`q` 是热流密度，单位 W/m²。负号表示热量沿温度降低的方向传递。

`def` 定义函数；`return` 返回计算结果。

```python
def heat_flux(k, dT, L):
    q = -k * dT / L
    return q

q = heat_flux(150, 20, 0.01)
print(q)
```

### 3.2 Анонимные функции: lambda (匿名函数)

`lambda` создаёт функцию из одного выражения. `map` применяет функцию к каждому элементу; `list` собирает результаты в список.

`lambda` 用一个表达式创建函数。`map` 对每个元素应用该函数；`list` 将结果组成列表。

```python
# Квадратная функция: f(x) = x**2
# 定义一个平方函数
square = lambda x: x ** 2
print(square(5))
```

```python
# Функция map применяет lambda к каждому элементу списка
# map 会把 lambda 函数应用到列表的每一个元素上
nums = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, nums))
print(squared)
```

## 4. Цикл (循环)
### 4.1 Цикл for（for 循环）
Используется для перебора объектов, из которых можно получать элементы по одному  
用于遍历那些能一个一个取出元素的对象（列表、元组、字符串、字典、集合）

#### 4.1.1 Перебор списка (遍历列表) :

```python
nums = [1, 2, 3, 4]
for num in nums:
    print(num)
```

#### 4.1.2 Перебор строки (遍历字符串):

```python
text = "abcd"
for char in text:
    print(char)
```

#### 4.1.3 Перебор словаря (遍历字典):

```python
dic = {'a': 1, 'b': 2, 'c': 3}
for key, value in dic.items():
    print(f"{key}: {value}")
```

#### 4.1.4 Использование range (range的使用):

```python
for i in range(4):
    print(i)
```

### 4.2 Цикл while（while 循环）
Выполняется, пока условие остаётся True. Если условие становится False, цикл завершается
(当条件为 True 时循环持续执行；一旦为 False，循环结束)

#### 4.2.1 Счётный цикл (计数循环)

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

#### 4.2.2 Условный цикл (条件循环)

```python
n = 10
while n > 0:
    print(n)
    n -= 2
```

### 4.3 Операторы управления циклами (循环控制语句)
#### 4.3.1 break — досрочно прерывает цикл (提前结束循环)
Используется, когда нужно немедленно выйти из цикла (用于立即跳出循环)

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

#### 4.3.2 continue — перейти к следующей итерации (跳过当前循环，继续下一轮)
Используется, когда нужно пропустить часть кода, но цикл продолжить (用于跳过本轮循环剩余代码，进入下一轮)

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

## 5. Дополнительные примеры функций (函数补充示例)

### 5.1 Среднее значение (平均值)

Список должен быть непустым.

输入列表不能为空。

```python
# Пользовательская функция: среднее значение списка
# 自定义函数：计算列表中所有元素的平均值
def average(lst):
    return sum(lst) / len(lst)

a = [2, 3, 1, -4, 12, 6]
print(average(a))
```

### 5.2 Число положительных элементов (正数元素的个数)

Считаем только значения, строго большие нуля.

仅统计严格大于 0 的元素。

```python
# Пользовательская функция: считаем количество положительных элементов
# 自定义函数：统计列表中 > 0 的元素个数
def count_positive(lst):
    count = 0
    for x in lst:
        if x > 0:
            count += 1
    return count

a = [2, 3, 1, -4, 12, 6, 0, -3]
print(count_positive(a))
```
