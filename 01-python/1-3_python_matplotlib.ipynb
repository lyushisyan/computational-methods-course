```python
import matplotlib.pyplot as plt
import numpy as np
```

## 1. Линейный график (折线图)

`plt.plot(x, y)` соединяет точки линией. Массивы `x` и `y` содержат по 100 значений.

`plt.plot(x, y)` 将点连接成曲线。`x` 和 `y` 各有 100 个值。

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

### 1.1 Несколько кривых и легенда (多条曲线与图例)

`label` задаёт подпись кривой, `legend()` показывает легенду.

`label` 设置曲线名称，`legend()` 显示图例。

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')

plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()
```

### 1.2 Размер рисунка и сохранение (图像尺寸与保存)

`figsize` задаёт размер в дюймах, `dpi` — число пикселей на дюйм. `bbox_inches="tight"` обрезает лишние поля. Сохраняйте рисунок перед `show()`.

`figsize` 设置英寸尺寸，`dpi` 设置每英寸像素数。`bbox_inches="tight"` 裁去多余边距。在 `show()` 之前保存图像。

```python
plt.figure(figsize=(6, 4), dpi=100,
           facecolor='white')

plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')

plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.savefig('plot.png', dpi=300,
            bbox_inches='tight')
plt.show()
```

### 1.3 Цвет, линии и подписи (颜色、线型与文字)

`linewidth` задаёт толщину линии, `fontsize` — размер текста, `tick_params` — оформление делений осей.

`linewidth` 设置线宽，`fontsize` 设置文字大小，`tick_params` 设置坐标轴刻度的样式。

```python
plt.figure(figsize=(6, 4), dpi=100,
           facecolor='white')
plt.plot(x, y1, label='sin(x)',
         color='red', linewidth=2)
plt.plot(x, y2, label='cos(x)',
         color='black', linestyle='--',
         linewidth=2)
plt.xlabel('x', fontsize=18)
plt.ylabel('y', fontsize=18)
plt.tick_params(labelsize=16)
plt.legend(fontsize=16)
plt.grid()
plt.savefig("plot.png", dpi=300,
            bbox_inches="tight")
plt.show()
```

## 2. Точки и линии с маркерами (散点与带标记的折线)

`scatter` показывает отдельные точки, `plot(..., marker="o")` соединяет точки линией и добавляет маркеры.

`scatter` 绘制独立的点，`plot(..., marker="o")` 在连接点的折线上添加圆形标记。

```python
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure(figsize=(6, 4), dpi=100)
plt.scatter(x, y1, color='blue',
            label='sin(x)')
plt.plot(x, y2, color='green',
         marker='o', linestyle='-',
         linewidth=2, label='cos(x)')
plt.xlabel('x', fontsize=18)
plt.ylabel('y', fontsize=18)
plt.tick_params(labelsize=16)
plt.legend(fontsize=14)
plt.grid()
plt.show()
```

## 3. Гистограмма (直方图)

`bins=30` задаёт 30 интервалов. Высота столбца — число значений в интервале. Случайная выборка меняется при каждом запуске.

`bins=30` 设置 30 个区间。柱高表示落入该区间的数据个数。每次运行的随机样本会不同。

```python
data = np.random.normal(
    loc=0, scale=1, size=1000)
plt.figure(figsize=(6, 4), dpi=100)
plt.hist(data, bins=30, 
    color='skyblue', 
    edgecolor='black',
    alpha=0.8, label='N(0,1)')
plt.xlabel('Значения', fontsize=16)
plt.ylabel('Частота', fontsize=16)
plt.tick_params(labelsize=14)
plt.legend(fontsize=14)
plt.tight_layout()
plt.show()
```

## 4. Двумерное поле (二维场)

### 4.1 Координатная сетка: meshgrid (坐标网格)

Три значения `x` и два значения `y` образуют шесть узлов. `X` хранит координаты по горизонтали, `Y` — по вертикали.

3 个 `x` 值和 2 个 `y` 值组成 6 个网格点。`X` 保存横坐标，`Y` 保存纵坐标。

```python
x = np.array([0, 1, 2])
y = np.array([0, 1])
X, Y = np.meshgrid(x, y)

print("X =")
print(X)
print("Y =")
print(Y)
```

### 4.2 Цветовая карта: contourf (二维云图)

Цвет показывает значение `Z`. `levels` управляет числом цветовых уровней, `colorbar` показывает цветовую шкалу.

颜色表示 `Z` 的值。`levels` 控制颜色分级，`colorbar` 显示色标。

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))
plt.figure(figsize=(6, 5))
contour = plt.contourf(
    X, Y, Z, levels=50,
    cmap='viridis')
cbar = plt.colorbar(contour)
cbar.set_label('z', fontsize=16)
plt.xlabel('x', fontsize=16)
plt.ylabel('y', fontsize=16)
plt.title('Contourf Plot', fontsize=18)
plt.tick_params(labelsize=14)
cbar.ax.tick_params(labelsize=14)
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 5. Трёхмерная поверхность (三维曲面)

Те же `X`, `Y`, `Z`: значение поля теперь задаёт высоту поверхности.

使用相同的 `X`、`Y`、`Z`，通过曲面高度展示场的数值。

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))
fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111,
    projection='3d')
surf = ax.plot_surface(X, Y, Z,
    cmap='viridis')
ax.set_xlabel('x', fontsize=14)
ax.set_ylabel('y', fontsize=14)
ax.set_zlabel('z', fontsize=14, labelpad=0)
plt.tight_layout(pad=2)
plt.show()
```

## 6. Несколько графиков: subplots (子图)

`fig` — весь рисунок, `axes[0]` и `axes[1]` — два графика. `sharex=True` задаёт общую шкалу по оси `x`.

`fig` 是整张图，`axes[0]` 和 `axes[1]` 是两个子图。`sharex=True` 使两个子图共用横坐标尺度。

```python
x = np.linspace(0, 2 * np.pi, 100)
fig, axes = plt.subplots(2, 1,
    figsize=(6, 5), sharex=True)
axes[0].plot(x, np.sin(x))
axes[0].set_ylabel('sin(x)',
                   fontsize=14)
axes[1].plot(x, np.cos(x), color='red')
axes[1].set_ylabel('cos(x)',
                   fontsize=14)
axes[1].set_xlabel('x', fontsize=14)
for ax in axes:
    ax.tick_params(labelsize=12)
    ax.grid()
fig.tight_layout()
plt.show()
```

## 7. Погрешности: errorbar (误差棒)

Учебные данные: температура и её погрешность. `yerr=dT` задаёт отрезки от `T-dT` до `T+dT`.

以下为温度及其误差的教学示例数据。`yerr=dT` 绘制从 `T-dT` 到 `T+dT` 的误差棒。

```python
t = np.array([0, 10, 20, 30, 40])
T = np.array([80, 65, 54, 46, 39])
dT = np.array([2, 3, 2, 2, 3])
plt.figure(figsize=(6, 4))
plt.errorbar(t, T, yerr=dT,
             fmt='o', capsize=5)
plt.xlabel('t, s', fontsize=16)
plt.ylabel('T, °C', fontsize=16)
plt.tick_params(labelsize=14)
plt.grid()
plt.tight_layout()
plt.show()
```

## 8. Логарифмическая шкала: semilogy (半对数图)

Здесь `dT` — разность температур тела и среды, а не погрешность. На логарифмической оси `y` экспоненциальное убывание выглядит как прямая. Значения должны быть положительными.

这里的 `dT` 表示物体与环境的温差，不是误差。指数衰减在对数纵轴下显示为直线，纵轴数值必须为正。

```python
t = np.linspace(0, 100, 200)
dT = 200 * np.exp(-0.05 * t)
fig, axes = plt.subplots(2, 1,
    figsize=(6, 5), sharex=True)
axes[0].plot(t, dT)
axes[1].semilogy(t, dT)
for ax in axes:
    ax.set_ylabel('T - T_env, K',
                  fontsize=12)
    ax.tick_params(labelsize=12)
    ax.grid()
axes[1].set_xlabel('t, s', fontsize=14)
fig.tight_layout()
plt.show()
```

## 9. Сохранение и загрузка численных данных (数值数据保存与读取)

### 9.1 Текстовый формат (文本格式)

`column_stack` объединяет `x` и `y` в два столбца. `savetxt` сохраняет таблицу, `loadtxt` читает её обратно.

`column_stack` 将 `x` 和 `y` 组成两列。`savetxt` 保存数据表，`loadtxt` 读取数据表。

```python
x = np.linspace(0, 1, 11)
y = x**2
data = np.column_stack((x, y))
np.savetxt("data.txt", data)
data = np.loadtxt("data.txt")
print(data[:3])
```

### 9.2 Формат NumPy: npy (NumPy 二进制格式)

Формат `.npy` сохраняет форму, тип и значения массива.

`.npy` 格式保存数组的形状、数据类型和数值。

```python
rng = np.random.default_rng(1)
A = rng.random((100, 100))
np.save("A.npy", A)
B = np.load("A.npy")
print(B.shape)
print(np.array_equal(A, B))
```
